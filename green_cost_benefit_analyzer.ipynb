```python
# ==========================================
# GREEN COST-BENEFIT ANALYZER - COMPARE 3 MACHINES
# ==========================================

# 1. Ask for the OLD machine's usage (stays the same for all)
old_kwh = float(input("Enter the OLD machine's yearly kWh usage: "))

# 2. Create empty lists to store data for 3 new machines
costs = []
savings = []
new_kwhs = []

# 3. Loop 3 times to get data for 3 different machines
for i in range(3):
    print(f"\n--- MACHINE {i+1} ---")
    cost = float(input("  Purchase price: $"))
    save = float(input("  Yearly energy savings vs old machine: $"))
    new_kwh = float(input("  New machine yearly kWh usage: "))

    # Add these to the lists
    costs.append(cost)
    savings.append(save)
    new_kwhs.append(new_kwh)

# 4. Calculate and display results for ALL 3 machines
print("\n" + "="*50)
print("COMPARATIVE ANALYSIS REPORT")
print("="*50)

emissions_factor = 0.85

for i in range(3):
    payback = costs[i] / savings[i]
    co2_saved = (old_kwh - new_kwhs[i]) * emissions_factor

    print(f"\nMACHINE {i+1}")
    print(f"   Cost: ${costs[i]:,.2f}")
    print(f"   Payback: {payback:.1f} years")
    print(f"   CO2 Saved: {co2_saved:.1f} lbs/year")
    print(f"   ROI per year: {(savings[i] / costs[i]) * 100:.1f}%")

print("\n" + "="*50)
print("RECOMMENDATION: The machine with the shortest payback is the best")
print("financial choice. The machine with the highest CO2 saved is the")
print("greenest choice.")
```
